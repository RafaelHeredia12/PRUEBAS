```python
# Ejercicio 3: Congruencial multiplicativo
a, m, x0 = 21, 32, 17

print("ALGORITMO CONGRUENCIAL MULTIPLICATIVO")
print(f"a = {a}, m = {m}, x0 = {x0}\n")
print(f"{'n':>3}  {'x_n':>4}  {'r_n':>8}")
print("-" * 19)

x = x0
n = 0
while True:
    x = (a * x) % m
    n += 1
    print(f"{n:>3}  {x:>4}  {x / (m - 1):>8.5f}")
    if x == x0:
        break

print("-" * 19)
print(f"Periodo = {n}")
```

```python
# Ejercicio 4: Congruencial mixto
import numpy as np
from scipy import stats

a, c, x0, m = 121, 553, 23, 177

# a) Generar enteros y periodo
x = x0
enteros = []
while True:
    x = (a * x + c) % m
    enteros.append(x)
    if x == x0:
        break

# b) Convertir a valores entre 0 y 1
r = np.array(enteros) / (m - 1)
n = len(r)

print("ALGORITMO CONGRUENCIAL MIXTO")
print(f"a = {a}, c = {c}, x0 = {x0}, m = {m}\n")
print("a) y b) Números generados")
print(f"{'i':>3}  {'x_i':>4}  {'r_i':>7}")
print("-" * 18)
for i in range(n):
    print(f"{i + 1:>3}  {enteros[i]:>4}  {r[i]:>7.4f}")
print("-" * 18)
print(f"Periodo = {n}\n")

# c) Pruebas con alfa = 0.05
alfa = 0.05

# Prueba de medias
media = r.mean()
z = stats.norm.ppf(1 - alfa / 2)
li = 0.5 - z / np.sqrt(12 * n)
ls = 0.5 + z / np.sqrt(12 * n)

print("c) PRUEBA DE MEDIAS (alfa = 0.05)")
print("   H0: media = 1/2")
print("   H1: media != 1/2")
print(f"   Media de r      = {media:.4f}")
print(f"   Límite inferior = {li:.4f}")
print(f"   Límite superior = {ls:.4f}")
if li <= media <= ls:
    print("   Conclusión: No se rechaza H0, la media es 1/2\n")
else:
    print("   Conclusión: Se rechaza H0, la media no es 1/2\n")

# Prueba de varianza
var = r.var(ddof=1)
li = stats.chi2.ppf(alfa / 2, n - 1) / (12 * (n - 1))
ls = stats.chi2.ppf(1 - alfa / 2, n - 1) / (12 * (n - 1))

print("c) PRUEBA DE VARIANZA (alfa = 0.05)")
print("   H0: varianza = 1/12")
print("   H1: varianza != 1/12")
print(f"   Varianza de r   = {var:.4f}")
print(f"   Límite inferior = {li:.4f}")
print(f"   Límite superior = {ls:.4f}")
if li <= var <= ls:
    print("   Conclusión: No se rechaza H0, la varianza es 1/12")
else:
    print("   Conclusión: Se rechaza H0, la varianza no es 1/12")
```
